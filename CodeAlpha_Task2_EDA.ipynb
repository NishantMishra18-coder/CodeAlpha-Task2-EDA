```python
import pandas as pd

df = pd.read_csv(
    "Sample - Superstore.csv",
    encoding="latin1"
)

df.head()
```

```python

# Step 3: Understand the Dataset

print("First 5 Rows:")
display(df.head())

print("\nDataset Shape (Rows, Columns):")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe(include="all"))
```

```python

# Step 4: Data Quality Check

print("Missing Values in Each Column:")
display(df.isnull().sum())

print("\nTotal Missing Values:")
print(df.isnull().sum().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nData Types:")
display(df.dtypes)
```

```python

# Step 5: Basic Sales Analysis

print("Total Sales:", df["Sales"].sum())
print("Total Profit:", df["Profit"].sum())

print("\nSales by Category:")
display(
    df.groupby("Category")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .reset_index()
)

# Visualize Sales by Category
df.groupby("Category")["Sales"].sum().sort_values().plot(
    kind="bar",
    title="Total Sales by Category",
    xlabel="Category",
    ylabel="Sales",
    figsize=(8, 5)
)

plt.tight_layout()
plt.show()
```

```python

# Step 6: Sales and Profit Analysis

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Sales and Profit by Category
category_analysis = df.groupby("Category")[["Sales", "Profit"]].sum()

print("Sales and Profit by Category:")
display(category_analysis)

# 2. Sales and Profit by Region
if "Region" in df.columns:
    region_analysis = df.groupby("Region")[["Sales", "Profit"]].sum()

    print("Sales and Profit by Region:")
    display(region_analysis)

# 3. Sales and Profit Visualization
category_analysis.plot(kind="bar", figsize=(9, 5))

plt.title("Sales and Profit by Category")
plt.xlabel("Category")
plt.ylabel("Amount")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

# Step 7: Monthly Sales Trend

date_col = next(
    (col for col in df.columns
     if col.lower() in ["order date", "order_date", "date"]),
    None
)

if date_col:
    df[date_col] = pd.to_datetime(df[date_col], errors="coerce")

    monthly_sales = (
        df.dropna(subset=[date_col])
          .set_index(date_col)["Sales"]
          .resample("ME")
          .sum()
    )

    monthly_sales.plot(figsize=(12, 5), marker="o")

    plt.title("Monthly Sales Trend")
    plt.xlabel("Month")
    plt.ylabel("Total Sales")
    plt.tight_layout()
    plt.show()
else:
    print("Date column nahi mila. Available columns:", df.columns.tolist())
```

```python

# Step 8: Top 10 Products by Sales

product_col = next(
    (col for col in df.columns
     if col.lower() in ["product name", "product_name", "product"]),
    None
)

if product_col:
    top_products = (
        df.groupby(product_col)["Sales"]
          .sum()
          .sort_values(ascending=False)
          .head(10)
    )

    display(top_products.reset_index())

    top_products.sort_values().plot(
        kind="barh", figsize=(10, 6)
    )

    plt.title("Top 10 Products by Sales")
    plt.xlabel("Total Sales")
    plt.tight_layout()
    plt.show()
else:
    print("Product column nahi mila.")
```

```python

# Step 9: Correlation Analysis

import matplotlib.pyplot as plt
import seaborn as sns

numeric_df = df.select_dtypes(include="number")
correlation = numeric_df.corr()

plt.figure(figsize=(10, 6))
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python

# Step 10: Outlier Detection using IQR

for col in ["Sales", "Profit"]:
    if col in df.columns:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1

        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR

        outliers = df[
            (df[col] < lower) | (df[col] > upper)
        ]

        print(f"\n{col}")
        print("Lower limit:", round(lower, 2))
        print("Upper limit:", round(upper, 2))
        print("Outlier count:", len(outliers))
```

```python

# Step 11: Business Summary

print("========== BUSINESS SUMMARY ==========")

print("\n1. Overall Sales:", round(df["Sales"].sum(), 2))
print("2. Overall Profit:", round(df["Profit"].sum(), 2))

if "Category" in df.columns:
    category_sales = df.groupby("Category")["Sales"].sum()
    print("\n3. Highest Sales Category:", category_sales.idxmax())
    print("4. Lowest Sales Category:", category_sales.idxmin())

if "Region" in df.columns:
    region_profit = df.groupby("Region")["Profit"].sum()
    print("\n5. Most Profitable Region:", region_profit.idxmax())
    print("6. Lowest Profit Region:", region_profit.idxmin())

print("\n======================================")
```
